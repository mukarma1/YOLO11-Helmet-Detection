```python
!pip install ultralytics opencv-python-headless -q

import cv2
from ultralytics import YOLO
from huggingface_hub import hf_hub_download
from google.colab import files

print("Models load ho rahe hain...")

# 1. Helmet Model (Custom)
model_helmet_path = hf_hub_download(repo_id="leeyunjai/yolo11-helmet", filename="helmet-11x.pt")
model_helmet = YOLO(model_helmet_path)

# 2. Standard YOLO Model (Bike aur Insaan ke liye)
# 'yolo11n.pt' standard model hai jo bike (class 3) aur person (class 0) pehchanta hai
model_standard = YOLO('yolo11n.pt')

video_path = "/content/video_2.mp4"
cap = cv2.VideoCapture(video_path)

# Video ki properties (width, height, fps)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Output video save karne ke liye
output_path = "/content/final_bikes_and_helmets.mp4"
out = cv2.VideoWriter(output_path, cv2.VideoWriter_fourcc(*'mp4v'), fps, (width, height))

print("Video process ho rahi hai... (Thoda waqt lagega)")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Model 1: Standard (Bike aur Person detect karega)
    # classes=[0, 3] ka matlab sirf Person aur Motorcycle detect karo
    res_standard = model_standard.predict(frame, conf=0.5, classes=[0, 3], verbose=False)[0]
    annotated_frame = res_standard.plot()

    # Model 2: Helmet (Helmet aur Face detect karega)
    res_helmet = model_helmet.predict(frame, conf=0.5, iou=0.45, verbose=False)[0]
    # Dono models ke boxes ko ek hi frame par draw karein
    annotated_frame = res_helmet.plot(img=annotated_frame)

    # Frame ko video mein likhein
    out.write(annotated_frame)

cap.release()
out.release()

print("✅ Video tayyar ho gayi hai!")
print("Download shuru ho raha hai...")
files.download(output_path)
```
